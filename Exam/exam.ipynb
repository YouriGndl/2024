**3.4 (6 pts)**: Build a Co-Author Network

Co-author networks help us visualize and quantify collaborations in academia. Such networks are critical for identifying influential researchers and their role in fostering connections within their communities. Construct a network where nodes represent authors and edges represent collaborations.

#### Instructions:
- Create a graph where: (5 pts)
  - Nodes: Each node represents a unique author, identified by their name in the authors field of the dataset.
  - Edges: There is an edge between two nodes if the corresponding authors have co-authored at least one paper. The weight of the edge represents the number of papers the two authors have co-authored together.
- After constructing the network, print the following information: (1 pt)
  - The total number of nodes (authors) in the network.
  - The total number of edges (collaborations) in the network.

Example:

For a paper authored by ["Alice", "Bob", "Charlie"], the graph should include:
- Nodes: Alice, Bob, Charlie.
- Edges: Alice-Bob, Alice-Charlie, Bob-Charlie, each with a weight of 1 (assuming no other collaborations between these pairs).


````python
# df.head()
# df_new = df['authors']

# my_list = []
# count = 0
# for i in df_new:
#     my_list = my_list + i
#     # count += 1
#     # if count > 10:
#     #     break

# my_liste_unique = []
# for index, name in enumerate(my_list):
#     if name not in my_list[index + 1:]:
#         my_liste_unique.append(name)

# print(my_liste_unique)
# #df_new.head()
###
def create_coauthor_network(df):
    # Define a function that takes a DataFrame 'df' as input
    # The DataFrame is expected to have a column called "authors",
    # where each row contains a list of authors for one paper

    G = nx.Graph()
    # Create an empty undirected graph using NetworkX
    # Nodes will be authors, edges will represent co-authorships

    for author_list in df["authors"]:
        # Loop over each paper's list of authors

        for i, author1 in enumerate(author_list): # enumerate is a Python function that returns both the index and the value of each element while iterating over a sequence.
            # Loop over authors in the current paper
            # 'i' is the index of author1 in the list

            for author2 in author_list[i + 1:]: # list[start : end] (ne end here)
                # Loop over all authors that come AFTER author1 in the list
                # This avoids:
                #  - self-loops (author with themselves)
                #  - duplicate pairs (A,B and B,A)

                if G.has_edge(author1, author2):
                    # Check if an edge already exists between author1 and author2
                    # This means they have co-authored a paper before

                    G[author1][author2]["weight"] += 1
                    # Increase the edge weight by 1
                    # The weight counts how many papers they co-authored

                else:
                    # If no edge exists yet between these two authors

                    G.add_edge(author1, author2, weight=1)
                    # Create a new edge between author1 and author2
                    # Initialize its weight to 1 (first co-authored paper)

    return G
    # Return the constructed co-author network graph


G = create_coauthor_network(df)
# Call the function to build the co-author graph from the DataFrame

print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")
# Print the number of nodes (authors) and edges (co-author relationships)

````
